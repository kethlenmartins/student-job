# Comparativo 2025 × 2026 — Pesquisa de Carreiras em Tech

Compara os dados do webscrapping Paulo (fev/2025) com o webscrapping versao3 (jul/2026).

**Fontes:**
- 2025: `webscrapping-2025/Códigos Webscrapping Paulo/data/scp_02.csv` e `scp_03.csv`
- 2026: `analytics/data/scp_02.csv` e `scp_03.csv`

```python
import pandas as pd
import numpy as np

PATH_25 = '../../../../webscrapping-2025/Códigos Webscrapping Paulo/data/'
PATH_26 = 'data/'

df25 = pd.read_csv(PATH_25 + 'scp_02.csv')
df26 = pd.read_csv(PATH_26 + 'scp_02.csv')
sk25 = pd.read_csv(PATH_25 + 'scp_03.csv')
sk26 = pd.read_csv(PATH_26 + 'scp_03.csv')

# Ambos os anos agora usam os mesmos nomes de coluna (sem /, - nas sk_)
SK_COLS = [c for c in sk26.columns if c.startswith('sk_')]
SK_COLS_COMUNS = [c for c in SK_COLS if c in sk25.columns]

print(f'2025: {len(df25):,} vagas  |  2026: {len(df26):,} vagas')
print(f'Skills comparáveis: {len(SK_COLS_COMUNS)}')
```

---
## 1. Volume e Distribuição por Área

```python
areas_25 = df25['Area'].value_counts()
areas_26 = df26['Area'].value_counts()

comp_area = pd.DataFrame({
    '2025 (n)': areas_25,
    '2025 (%)': (areas_25 / len(df25) * 100).round(1),
    '2026 (n)': areas_26,
    '2026 (%)': (areas_26 / len(df26) * 100).round(1),
}).fillna(0)

comp_area['Δ pp'] = (comp_area['2026 (%)'] - comp_area['2025 (%)']).round(1)
comp_area = comp_area.sort_values('2025 (n)', ascending=False)
comp_area[['2025 (n)', '2025 (%)', '2026 (n)', '2026 (%)', 'Δ pp']] = comp_area[['2025 (n)', '2025 (%)', '2026 (n)', '2026 (%)', 'Δ pp']].astype({'2025 (n)': int, '2026 (n)': int})

print(f'Total  |  2025: {len(df25):,}  |  2026: {len(df26):,}  |  Δ: {len(df26)-len(df25):+,}')
print()
display(comp_area)
```

---
## 2. Distribuição por Nível de Senioridade

```python
nivel_25 = df25['Nivel'].value_counts()
nivel_26 = df26['Nivel'].value_counts()

comp_nivel = pd.DataFrame({
    '2025 (n)': nivel_25,
    '2025 (%)': (nivel_25 / len(df25) * 100).round(1),
    '2026 (n)': nivel_26,
    '2026 (%)': (nivel_26 / len(df26) * 100).round(1),
}).fillna(0)

comp_nivel['Δ pp'] = (comp_nivel['2026 (%)'] - comp_nivel['2025 (%)']).round(1)
ordem = ['Nível Júnior', 'Nível Pleno / Sênior', 'Nível Liderança', 'n/f']
comp_nivel = comp_nivel.reindex([n for n in ordem if n in comp_nivel.index])

display(comp_nivel)
```

---
## 3. Distribuição Geográfica — Top 10 Estados

```python
estados_25 = df25[df25['Local Estado'] != 'n/f']['Local Estado'].value_counts().head(10)
estados_26 = df26[df26['Local Estado'] != 'n/f']['Local Estado'].value_counts().head(10)

todos_estados = sorted(set(estados_25.index) | set(estados_26.index))

comp_geo = pd.DataFrame({
    '2025 (n)': estados_25.reindex(todos_estados, fill_value=0),
    '2025 (%)': (estados_25.reindex(todos_estados, fill_value=0) / len(df25) * 100).round(1),
    '2026 (n)': estados_26.reindex(todos_estados, fill_value=0),
    '2026 (%)': (estados_26.reindex(todos_estados, fill_value=0) / len(df26) * 100).round(1),
})
comp_geo['Δ pp'] = (comp_geo['2026 (%)'] - comp_geo['2025 (%)']).round(1)
comp_geo = comp_geo.sort_values('2025 (n)', ascending=False)

display(comp_geo)
```

### 3.1 Tipo de trabalho (presencial / remoto / híbrido)

```python
tipo_25 = df25[df25['Local Tipo'].isin(['nacional', 'remoto', 'hibrido'])]['Local Tipo'].value_counts()
tipo_26 = df26[df26['Local Tipo'].isin(['nacional', 'remoto', 'hibrido'])]['Local Tipo'].value_counts()

comp_tipo = pd.DataFrame({
    '2025 (%)': (tipo_25 / tipo_25.sum() * 100).round(1),
    '2026 (%)': (tipo_26 / tipo_26.sum() * 100).round(1),
}).fillna(0)
comp_tipo['Δ pp'] = (comp_tipo['2026 (%)'] - comp_tipo['2025 (%)']).round(1)

display(comp_tipo)
```

---
## 4. Top Skills Técnicas (Hard Skills + Linguagens)

```python
# Carrega metadados de skills do criterios 2026
df_crit = pd.read_excel('data/criterios.xlsx', sheet_name='skills')

def normalize_col(text):
    import re
    from unidecode import unidecode
    t = str(text).lower().strip()
    for k, v in [('c#','csharp'),('c++','cplusplus'),('.net','dotnet'),('node.js','nodejs')]:
        t = t.replace(k, v)
    t = unidecode(t)
    t = re.sub(r'[^a-z0-9]+', '_', t).strip('_')
    return f'sk_{t}'

df_crit['col_name'] = df_crit['Skill'].apply(normalize_col)
tech_tipos = ['hard skill', 'programação']
meta_tech = df_crit[df_crit['Tipo'].str.lower().isin(tech_tipos)].copy()
cols_tech = [c for c in meta_tech['col_name'] if c in SK_COLS_COMUNS]

soma_25 = sk25[cols_tech].sum().sort_values(ascending=False)
soma_26 = sk26[cols_tech].sum().sort_values(ascending=False)

perc_25 = (soma_25 / len(sk25) * 100).round(1)
perc_26 = (soma_26 / len(sk26) * 100).round(1)

# top 20 pela média dos dois anos
media = ((perc_25 + perc_26) / 2).sort_values(ascending=False).head(20)

skill_label = df_crit.set_index('col_name')['Skill']

comp_tech = pd.DataFrame({
    'Skill': [skill_label.get(c, c) for c in media.index],
    '2025 (%)': perc_25.reindex(media.index).values,
    '2026 (%)': perc_26.reindex(media.index).values,
}, index=range(1, 21))
comp_tech.index.name = 'Rank'
comp_tech['Δ pp'] = (comp_tech['2026 (%)'] - comp_tech['2025 (%)']).round(1)
comp_tech['Tendência'] = comp_tech['Δ pp'].apply(lambda x: '↑' if x > 0 else ('↓' if x < 0 else '='))

print('Top 20 Hard Skills — ordenadas pela média dos dois anos')
display(comp_tech)
```

---
## 5. Top Soft Skills

```python
meta_soft = df_crit[df_crit['Tipo'].str.lower() == 'soft skill'].copy()
cols_soft = [c for c in meta_soft['col_name'] if c in SK_COLS_COMUNS]

soma_s25 = sk25[cols_soft].sum().sort_values(ascending=False)
soma_s26 = sk26[cols_soft].sum().sort_values(ascending=False)

perc_s25 = (soma_s25 / len(sk25) * 100).round(1)
perc_s26 = (soma_s26 / len(sk26) * 100).round(1)

media_soft = ((perc_s25 + perc_s26) / 2).sort_values(ascending=False)

comp_soft = pd.DataFrame({
    'Skill': [skill_label.get(c, c) for c in media_soft.index],
    '2025 (%)': perc_s25.reindex(media_soft.index).values,
    '2026 (%)': perc_s26.reindex(media_soft.index).values,
}, index=range(1, len(media_soft)+1))
comp_soft.index.name = 'Rank'
comp_soft['Δ pp'] = (comp_soft['2026 (%)'] - comp_soft['2025 (%)']).round(1)
comp_soft['Tendência'] = comp_soft['Δ pp'].apply(lambda x: '↑' if x > 0 else ('↓' if x < 0 else '='))

display(comp_soft)
```

## 6. Top Skills por Área — comparativo

```python
areas = sorted(df26['Area'].dropna().unique())
TOP_N = 10

for area in areas:
    idx25 = sk25.index[df25['Area'] == area]
    idx26 = sk26.index[df26['Area'] == area]

    if len(idx25) == 0 or len(idx26) == 0:
        continue

    p25_full = (sk25.loc[idx25, SK_COLS_COMUNS].sum() / len(idx25) * 100)
    p26_full = (sk26.loc[idx26, SK_COLS_COMUNS].sum() / len(idx26) * 100)

    top25 = p25_full.sort_values(ascending=False).head(TOP_N).index
    top26 = p26_full.sort_values(ascending=False).head(TOP_N).index
    top_skills = pd.Index(list(top25) + list(top26)).unique()

    media_area = ((p25_full.reindex(top_skills, fill_value=0) + p26_full.reindex(top_skills, fill_value=0)) / 2).sort_values(ascending=False).head(TOP_N)

    res = pd.DataFrame({
        'Skill': [skill_label.get(c, c) for c in media_area.index],
        '2025 (%)': p25_full.reindex(media_area.index, fill_value=0).round(1).values,
        '2026 (%)': p26_full.reindex(media_area.index, fill_value=0).round(1).values,
    }, index=range(1, len(media_area)+1))
    res.index.name = 'Rank'
    res['Δ pp'] = (res['2026 (%)'] - res['2025 (%)']).round(1)
    res['Tendência'] = res['Δ pp'].apply(lambda x: '↑' if x > 0 else ('↓' if x < 0 else '='))

    print(f'\n{"="*60}')
    print(f'  {area}  |  2025: {len(idx25):,} vagas  →  2026: {len(idx26):,} vagas')
    print(f'{"="*60}')
    display(res)
```

## 7. Salário por Nível

```python
import re
from unidecode import unidecode

# ── 2025: carrega xlsx já processado (tem nivel) ──────────────────────────
sal_25 = pd.read_excel(PATH_25 + 'analise_salario_por_nivel.xlsx', index_col=0)
print('=== 2025 — Salário médio por nível (R$ mil) ===')
display(sal_25)

# ── 2026: parseia salário + nível diretamente do scp_01 ──────────────────
def parse_faixa(text):
    if not isinstance(text, str) or text.strip().upper() in ['NF', '', 'N/E', 'N/F']:
        return None
    text = text.replace('\xa0', ' ')
    nums = re.findall(r'R\$\s*([\d\.]+)', text)
    if not nums:
        return None
    try:
        val_k = float(nums[0].replace('.', '')) / 1000
        if val_k < 2:    return '|- 2'
        elif val_k < 5:  return '2 |- 5'
        elif val_k < 10: return '5 |- 10'
        elif val_k < 30: return '10 |- 30'
        else:            return '30 |-'
    except Exception:
        return None

df26_s1 = pd.read_csv(PATH_26 + 'scp_01.csv')

# Classifica Nivel inline usando os mesmos critérios do scp_02
df_nivel_cargo = pd.read_excel(PATH_26 + 'criterios.xlsx', sheet_name='nivel_cargo')
df_nivel_cargo = df_nivel_cargo.sort_values(by='Prioridade')

def match_nivel(text):
    if text == '' or pd.isna(text):
        return 'n/e'
    res = 'n/f'
    rows = df_nivel_cargo[df_nivel_cargo['Cargo'].apply(lambda k: f' {k} ' in f' {text} ')]
    if rows.shape[0] > 0:
        for _, row in rows.iterrows():
            for column in df_nivel_cargo.columns[2:]:
                if row[column] == 1:
                    res = column
                    break
    return res

df26_s1['Nivel'] = df26_s1['Cargo'].apply(match_nivel)
df26_s1['Faixa'] = df26_s1['Salario'].apply(parse_faixa)
df26_sal_ok = df26_s1[df26_s1['Faixa'].notna()].copy()

n_total_26 = len(df26_s1)
n_com_sal_26 = len(df26_sal_ok)
print(f'\n2026 — vagas com salário: {n_com_sal_26:,} ({n_com_sal_26/n_total_26*100:.1f}%)')

faixa_ordem = {'|- 2': 1, '2 |- 5': 3.5, '5 |- 10': 7.5, '10 |- 30': 20, '30 |-': 35}
df26_sal_ok['Salario_Medio'] = df26_sal_ok['Faixa'].map(faixa_ordem)

sal_26_geral = df26_sal_ok.groupby('Faixa')['Salario_Medio'].agg(
    Quantidade='count'
).reindex(['|- 2','2 |- 5','5 |- 10','10 |- 30','30 |-'])
sal_26_geral['% vagas'] = (sal_26_geral['Quantidade'] / n_com_sal_26 * 100).round(1)

print('\n=== 2026 — Distribuição por faixa salarial (R$ mil) ===')
display(sal_26_geral)

# ── Comparativo por nível ─────────────────────────────────────────────────
sal_26_nivel = df26_sal_ok.groupby('Nivel')['Salario_Medio'].agg(
    Média='mean', Quantidade='count'
).round(2)

niveis_comuns = sal_25.index.intersection(sal_26_nivel.index)
if len(niveis_comuns) > 0 and 'Média' in sal_25.columns:
    comp_sal = pd.DataFrame({
        'Média 2025 (R$ mil)': sal_25.loc[niveis_comuns, 'Média'],
        'Média 2026 (R$ mil)': sal_26_nivel.loc[niveis_comuns, 'Média'],
    })
    comp_sal['Δ'] = (comp_sal['Média 2026 (R$ mil)'] - comp_sal['Média 2025 (R$ mil)']).round(2)
    ordem_nivel = ['Nível Júnior', 'Nível Pleno / Sênior', 'Nível Liderança', 'n/f']
    comp_sal = comp_sal.reindex([n for n in ordem_nivel if n in comp_sal.index])
    print('\n=== Comparativo — ponto médio salarial por nível (R$ mil) ===')
    display(comp_sal)
else:
    print('(sem níveis em comum para comparar)')
```

## 8. Vagas com Foco em IA

```python
ia25 = sk25['IA'].sum() if 'IA' in sk25.columns else 0
ia26 = sk26['IA'].sum() if 'IA' in sk26.columns else 0

perc_ia25 = ia25 / len(sk25) * 100
perc_ia26 = ia26 / len(sk26) * 100

print(f'Vagas com foco em IA:')
print(f'  2025: {ia25:,} ({perc_ia25:.1f}% do total)')
print(f'  2026: {ia26:,} ({perc_ia26:.1f}% do total)')
print(f'  Δ pp: {perc_ia26 - perc_ia25:+.1f} pp')
print()

# IA por área
if 'IA' in sk26.columns:
    ia_area_25 = sk25.groupby(df25['Area'])['IA'].mean() * 100
    ia_area_26 = sk26.groupby(df26['Area'])['IA'].mean() * 100

    comp_ia_area = pd.DataFrame({
        '2025 (%)': ia_area_25.round(1),
        '2026 (%)': ia_area_26.round(1),
    }).fillna(0)
    comp_ia_area['Δ pp'] = (comp_ia_area['2026 (%)'] - comp_ia_area['2025 (%)']).round(1)
    comp_ia_area = comp_ia_area.sort_values('2026 (%)', ascending=False)

    print('Vagas com foco em IA por área:')
    display(comp_ia_area)
```
